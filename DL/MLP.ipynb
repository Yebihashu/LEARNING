```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
words = open('names.txt', 'r').read().splitlines()
len(words)
print(words[:8])
```

```python
# build vocabulary of unique characters, and map each character to an integer and vice versa
chars = sorted(list(set(''.join(words))))
print(chars)
stoi = { ch:(i+1) for i,ch in enumerate(chars) }
stoi['.'] = 0
itos = { i:ch for ch,i in stoi.items() }
print(stoi)
print(itos)
```

```python
# build the data set
X, Y = [], []
block_size = 3
for w in words[:5]:
    print(w)
    context = [0] * block_size
    for ch in w + '.':
        ix = stoi[ch]
        X.append(context)
        Y.append(ix)
        print(''.join([itos[i] for i in context]), '->', itos[ix])
        context = context[1:] + [ix]

X = torch.tensor(X)
Y = torch.tensor(Y)
print(X.shape, Y.shape)



        
```
